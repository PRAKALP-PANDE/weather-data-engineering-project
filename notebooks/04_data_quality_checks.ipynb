###Data Quality Layer

```python
silver_df = spark.table(
    "weather_silver"
)
```

```python
silver_df.display()
```

```python
from pyspark.sql.functions import *
silver_df.filter(
    (col("temperature") < -50)
    | (col("temperature") > 60)
).count()
```

```python
silver_df.filter(
    (col("humidity") < 0)
    | (col("humidity") > 100)
).count()
```

```python
silver_df.filter(
    col("city").isNull()
).count()
```

```python
temperature_failures = silver_df.filter(
    (col("temperature") < -50) |
    (col("temperature") > 60)
).count()

humidity_failures = silver_df.filter(
    (col("humidity") < 0) |
    (col("humidity") > 100)
).count()

city_failures = silver_df.filter(
    col("city").isNull()
).count()
```

```python
from pyspark.sql import Row

quality_results = [
    Row(check_name="temperature_range",
        failed_records=temperature_failures),

    Row(check_name="humidity_range",
        failed_records=humidity_failures),

    Row(check_name="city_not_null",
        failed_records=city_failures)
]

quality_df = spark.createDataFrame(
    quality_results
)

display(quality_df)
```
