```python
from pyspark.sql.functions import col, max

silver_df = spark.table("weather_silver")

latest_timestamp = (
    silver_df
    .agg(max("ingestion_timestamp"))
    .collect()[0][0]
)

print("Latest Processed Timestamp:", latest_timestamp)
```

```python
new_records_df = (
    silver_df
    .filter(
        col("ingestion_timestamp") >= latest_timestamp
    )
)

display(new_records_df)
```

```python
from pyspark.sql.functions import count

city_counts = (
    silver_df
    .groupBy("city")
    .agg(
        count("*").alias("record_count")
    )
)

display(city_counts)
```
