```python
bronze_df = (
    spark.read
    .option("multiLine", "true")
    .json("/Volumes/workspace/default/weather_raw")
)

display(bronze_df)
```

```python
bronze_df.printSchema()
```

```python
from pyspark.sql.functions import col

silver_df = bronze_df.select(
    col("name").alias("city"),
    col("sys.country").alias("country"),

    col("main.temp").alias("temperature"),
    col("main.feels_like").alias("feels_like"),

    col("main.temp_min").alias("temp_min"),
    col("main.temp_max").alias("temp_max"),

    col("main.humidity").alias("humidity"),
    col("main.pressure").alias("pressure"),

    col("wind.speed").alias("wind_speed"),
    col("wind.deg").alias("wind_degree"),

    col("visibility"),

    col("clouds.all").alias("cloudiness"),

    col("weather")[0]["main"].alias("weather_main"),
    col("weather")[0]["description"].alias("weather_description"),

    col("sys.sunrise").alias("sunrise"),
    col("sys.sunset").alias("sunset"),

    col("ingestion_timestamp")
)

display(silver_df)
```

```python
silver_df.write.mode("overwrite").parquet(
    "/Volumes/workspace/default/weather_raw/silver"
)
```

```python
from pyspark.sql.functions import (
    col,
    from_unixtime,
    to_timestamp
)

silver_df = bronze_df.select(
    col("name").alias("city"),
    col("sys.country").alias("country"),

    col("main.temp").alias("temperature"),
    col("main.feels_like").alias("feels_like"),

    col("main.temp_min").alias("temp_min"),
    col("main.temp_max").alias("temp_max"),

    col("main.humidity").alias("humidity"),
    col("main.pressure").alias("pressure"),

    col("wind.speed").alias("wind_speed"),
    col("wind.deg").alias("wind_degree"),

    col("visibility"),

    col("clouds.all").alias("cloudiness"),

    col("weather")[0]["main"].alias("weather_main"),
    col("weather")[0]["description"].alias("weather_description"),

    to_timestamp(
        from_unixtime(col("sys.sunrise"))
    ).alias("sunrise"),

    to_timestamp(
        from_unixtime(col("sys.sunset"))
    ).alias("sunset"),

    to_timestamp(
        col("ingestion_timestamp")
    ).alias("ingestion_timestamp")
)

display(silver_df)
```

###Phase 6: Build Gold Layer

Average Temperature by City

```python
from pyspark.sql.functions import avg

gold_avg_temp = (
    silver_df
    .groupBy("city")
    .agg(
        avg("temperature")
        .alias("avg_temperature")
    )
)

display(gold_avg_temp)
```

Highest Temperature Recorded

```python
from pyspark.sql.functions import max

gold_max_temp = (
    silver_df
    .groupBy("city")
    .agg(
        max("temperature")
        .alias("max_temperature")
    )
)

display(gold_max_temp)
```
