```python
silver_df = spark.table(
    "weather_silver"
)
```

```python
from pyspark.sql.functions import avg

gold_avg_temp = (
    silver_df
    .groupBy("city")
    .agg(
        avg("temperature")
        .alias("avg_temperature")
    )
)

display(gold_avg_temp)
```

```python
from pyspark.sql.functions import avg

gold_avg_temp = (
    silver_df
    .groupBy("city")
    .agg(
        avg("temperature")
        .alias("avg_temperature")
    )
)
gold_avg_temp.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("weather_gold_avg_temp")
```

```python
from pyspark.sql.functions import avg

gold_avg_humidity = (
    silver_df
    .groupBy("city")
    .agg(
        avg("humidity")
        .alias('avg_humidity')
    )
)

gold_avg_humidity.write \
    .format("delta") \
        .mode("overwrite") \
            .saveAsTable("weather_gold_avg_humidity")
```

```python
from pyspark.sql.functions import *

gold_max_temp = (
    silver_df
    .groupBy("city")
    .agg(
        max("temperature")
        .alias("max_temperature")
    )
)

gold_max_temp.write \
    .format("delta") \
        .mode("overwrite") \
            .saveAsTable("weather_gold_max_temp")
```
